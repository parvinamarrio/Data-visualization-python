## Libraries

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

## Load data

```python
cyclones = pd.read_csv("Chennai cyclones.csv")
```

```python
cyclones
```

## Basic line plot

```python
plt.plot(cyclones['Year'], cyclones['Fatalities'])
plt.xlabel('Year')
plt.ylabel('death rate')
plt.title('Chennai floods')
plt.show()
```

## Plot beautifier(colours,marker,linewidth,grid,linestyle,markersize)

```python
plt.plot(cyclones['Year'], cyclones['Fatalities'], color='red', marker = 'o', linewidth=5, linestyle='dotted', markersize='15')
plt.grid(True)
plt.xlabel('Year')
plt.ylabel('death rate')
plt.title('Chennai floods')
plt.show()
```

## Multiple plots

```python
fig = plt.figure(figsize=(25,15))

ax1 = fig.add_subplot(2,2,1)

plt.plot(cyclones['Year'], cyclones['Fatalities'], color='red', marker = 'o', linewidth=5, linestyle='dotted', markersize='15')
plt.grid(True)
plt.xlabel('Year')
plt.ylabel('death rate')
plt.title('Chennai floods')

ax2 = fig.add_subplot(2,2,2)

plt.plot(cyclones['Cyclone Name'], cyclones['Fatalities'], color='cyan', marker = 'o', linewidth=5, linestyle='solid', markersize='15')
plt.grid(True)
plt.xlabel('name of cyclone')
plt.ylabel('death rate')
plt.title('Chennai floods')

ax3 = fig.add_subplot(2,2,3)

plt.plot(cyclones['Formed'], cyclones['Fatalities'], color='green', marker = 'o', linewidth=5, linestyle='dashed', markersize='15')
plt.grid(True)
plt.xlabel('cyclone birth')
plt.ylabel('death rate')
plt.title('Chennai floods')

ax4 = fig.add_subplot(2,2,4)

plt.plot(cyclones['Dissipated'], cyclones['Fatalities'], color='blue', marker = 'o', linewidth=5, linestyle='dashdot', markersize='15')
plt.grid(True)
plt.xlabel('cyclone death')
plt.ylabel('death rate')
plt.title('Chennai floods')
```

## Bar plot

```python
def convert_damages(damages):
    endresult = []
    for i in damages:
        values = i.split()
        # print(values)
        num = values[0][1:]
        print(type(num))
        if values[1] == 'million':
            result = float(num) * 1000000
        else:
            result = float(num) * 1000000000
        endresult.append(result)
    return endresult   
    
```

```python
damage_converted = convert_damages(cyclones['Damage'])
```

```python
cyclones['Damage']
```

```python
damage_converted
```

```python
cyclones['Damage_converted'] = damage_converted
```

```python
cyclones
```

```python
fig = plt.figure(figsize=(25,10))
ax = fig.add_subplot(1,1,1)

ax.xaxis.label.set_color('green')
ax.yaxis.label.set_color('blue')

ax.tick_params(axis='x', colors='red')
ax.tick_params(axis='y', colors='black')


plt.bar(cyclones['Cyclone Name'], cyclones['Damage_converted'])
plt.xlabel('name of cyclone')
plt.xticks(rotation=40)
plt.ylabel('damages')
plt.title('which cyclone is expensive?')

plt.show()
```
